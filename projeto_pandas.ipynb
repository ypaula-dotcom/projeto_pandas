## Perguntas sobre o dataset

### 1. Quantos registros há para cada gênero?

```python
dados["Gender"].value_counts()
```

### 2. Quais são os registros do gênero feminino?

```python
dados[dados["Gender"] == "F"]
```

### 3. Quais são os registros do gênero masculino?

```python
dados[dados["Gender"] == "M"]
```

### 4. Qual é a média da quantidade (Count)?

```python
dados["Count"].mean()
```

### 5. Qual é o nome com a maior quantidade de ocorrências?


```python
dados.sort_values("Count", ascending=False).head(1)
```

### 6. Quais são os 10 nomes da categoria F com mais ocorrências?

```python
dados[dados["Gender"] == "F"].sort_values("Count", ascending=False).head(10)
```

### 7. Qual é a média de ocorrências dos nomes da categoria F?

```python
dados[dados["Gender"] == "F"]["Count"].mean()
```

### 8. Quais são os 10 nomes da categoria M com mais ocorrências?

```python
dados[dados["Gender"] == "M"].sort_values("Count", ascending=False).head(10)
```

### 9. Qual é o percentual de registros de cada categoria?

```python
dados["Gender"].value_counts(normalize=True) * 100
```

### 10. Quantos registros possuem mais de 100.000 ocorrências em cada categoria?

```python
dados[dados["Count"] > 100000]["Gender"].value_counts()
```
